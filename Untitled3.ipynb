```python
import sqlite3
import matplotlib.pyplot as plt

# Conectar a la base de datos SQLite
db_path = "C:\\googleMapsVS\\Google-Maps\\resultados_programa.db"
conn = sqlite3.connect(db_path)
c = conn.cursor()

# Consulta para recuperar datos de la base de datos
query = '''
SELECT nameProblem, algorithm, nodesGenerated
FROM resultados
'''

# Ejecutar la consulta y obtener todos los resultados
c.execute(query)
data = c.fetchall()

# Cerrar la conexión a la base de datos
conn.close()

# Preparar los datos para la gráfica
problems = {}
for row in data:
    nameProblem, algorithm, nodesGenerated = row
    if nameProblem not in problems:
        problems[nameProblem] = {}
    if algorithm not in problems[nameProblem]:
        problems[nameProblem][algorithm] = []
    problems[nameProblem][algorithm].append(nodesGenerated)

# Crear la gráfica
plt.figure(figsize=(10, 6))

for problem in problems:
    for algorithm in problems[problem]:
        plt.scatter([problem] * len(problems[problem][algorithm]), problems[problem][algorithm], label=algorithm)

plt.xlabel('Problema')
plt.ylabel('Nodos Generados')
plt.title('Nodos Generados por Estrategia para Cada Problema')
plt.legend(title='Algoritmo', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
